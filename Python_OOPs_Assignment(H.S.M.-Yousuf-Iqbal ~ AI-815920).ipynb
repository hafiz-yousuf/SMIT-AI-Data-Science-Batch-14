**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

```python
class Student:
    def __init__(self,name,roll_number):
        self.__name=name   #private variables
        self.__roll_number=roll_number
        
    #getter method
    def get_name(self):
        return self.__name

    def get_roll_number(self):
        return self.__roll_number

    #setter method
    def set_name(self,new_name):
        self.__name=new_name
        

    def set_roll_number(self,new_roll_number):
        self.__roll_number=new_roll_number


std1=Student("Ali",11)


print(f"Name : {std1.get_name()}")
print(f"Roll Number : {std1.get_roll_number()}")


std1.set_name("Saira")
std1.set_roll_number(12)

        
print(f"Updated Name : {std1.get_name()}")
print(f"Updated Roll Number : {std1.get_roll_number()}")       
```

Q2: Aik BankAccount class banayein jismein __balance private variable ho. Aik deposit() aur aik withdraw() method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.


```python
class BankAccount:
    def __init__(self,balance):
        self.__balance=balance

    def deposit(self,amount):
        self.__balance+=amount
        print(f"Rs.{amount} has been deposited, Your current account balance is {self.__balance}")

    def withdraw(self,amount):
        if amount>self.__balance:
            print("Insufficient balance")
        else:
            self.__balance-=amount
            print(f"Rs.{amount} has been withdrawn, Your current account balance is {self.__balance}")


my_account=BankAccount(12000)

my_account.deposit(2000)

my_account.withdraw(10000)

my_account.withdraw(5000)
```

**Q3:** Aik class banayein `Employee` jismein salary private ho. Agar koi salary ko minus (negative) value set karne ki koshish kare, toh setter method 'Invalid Salary' ka error message show kare.

```python
class Employee:
    def __init__(self,salary):
        self.set_salary(salary)

    def get_salary(self):
        return self.__salary

    def set_salary(self,new_salary):
        if new_salary<0:
            print("Invalid Salary")
            self.__salary=0
        else:
            self.__salary=new_salary


emp1=Employee(65000)
print(f"Salary : {emp1.get_salary()}")

emp1.set_salary(-65000)

```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
class Shape:
    def __init__(self,color):
        self.color=color
    def show_color(self):
        print(f"Color : {self.color}")

class Circle(Shape):
    def __init__(self,color,diameter):
        super().__init__(color)
        self.diameter=diameter

    def show_circle(self):
        print(f"Circle Diameter : {self.diameter}")


class Square(Shape):
    def __init__(self,color,sides):
        super().__init__(color)
        self.sides=sides

    def show_square(self):
        print(f"Square Side : {self.sides}")


c1=Circle("Green",3.5)
s1=Square("Yellow",4)

c1.show_color()
c1.show_circle()

s1.show_color()
s1.show_square()
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
class Book:
    def __init__(self,pages):
        self.pages=pages

    def __add__(self,other):
        total_pages=self.pages+other.pages
        return total_pages

bokk1=Book(50)
book2=Book(100)

no_of_pages=bokk1+book2

print(f"No of pages : {no_of_pages}")
```

**Q15:** Kya Python default oop se Method Overloading (same method name with different arguments) support karta hai? Agar nahi, toh hum same name ke method ko multiple arguments ke sath kaise handle karte hain? Code likhein.

```python
class Calculator:
    def add(self, *numbers):
        total = 0

        for num in numbers:
            total += num

        return total


calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
print(calc.add(5, 10, 15, 20))
```

**Q16:** `abc` module ka use karte hue aik abstract class `Computer` banayein jisme abstract method `process()` ho. Is class ka object banane ki koshish karein aur dekhein kya hota hai.

```python
from abc import ABC, abstractmethod


class Computer(ABC):

    @abstractmethod
    def process(self):
        pass


class Laptop(Computer):

    def process(self):
        print("Laptop is processing data")


# Child class ka object possible, parent class ka possible nh chun k uska method adhura h
laptop1 = Laptop()
laptop1.process()
```

**Q17:** Q16 wali `Computer` class ko inherit karke do child classes `Laptop` aur `Desktop` banayein aur dono mein `process()` method ko mandatory implement karein.

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
from abc import ABC,abstractmethod

class Car(ABC):
    def start_engine(self):
        print("The engine is starting")

        
    @abstractmethod
    def drive(self):
        pass


class Toyota(Car):
    def drive (self):
        print("Toyota is best car")

car1=Toyota()

car1.start_engine()
car1.drive()

```

**Q19:** Abstraction aur Encapsulation ke darmiyan kya wazeh farq (main difference) hai? Dono ke concepts ko short statements mei define karein.

Abstraction:
It is the process of hiding complex internal details and showing only the essential features.
Encapsulation:
It is the process of binding (wrapping) data (variables) and methods (functions) together into a single unit (a class) and restricting direct access to them.


Main difference:
Abstraction focuses on hiding complexity.
Encapsulation focuses on protecting data.


**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
from abc import ABC,abstractmethod

class Payment:

    @abstractmethod
    def pay(self,amount):
        pass

class CreditCardPayment(Payment):
    def pay(self,amount):
        print(f"paid {amount} using credit card")

class UPIPayment(Payment):
    def pay(self,amount):
        print(f"paid {amount} using UPI payment")


payment1=CreditCardPayment()
payment2=UPIPayment()

payment1.pay(5000)
payment2.pay(2000)
```
